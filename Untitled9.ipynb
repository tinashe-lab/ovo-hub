```python
#this imports the python data library
import pandas as pd

#this is to load the csv file
df=pd.read_csv('climate_data.csv')

#this is to remove certain columns
df.drop(columns=['valueType','timeCoverage','time_detail',], inplace=True)

#this is to remove the first 5 values
print(df.head())

#this is to show the data types in the
print("\n These are the data types of the data ")
print(df.dtypes)

#Thi shows how many missing values are there
print("\n This the amount of missing values")
print(df.isnull().sum())

#For the United Kingdom of Great Britain and Northern Ireland what was the proportionof renewable energy in 2022?
print("\n This is the proportion of renewable energy in 2022")
uk = df[df['geoAreaName'] == 'United Kingdom of Great Britain and Northern Ireland']
print(uk[uk['timePeriodStart'] == 2022]['EG_FEC_RNEW'])

#How many Megatonnes of CO2equivalent greenhouse gases did Mauritius emit in 2020?
print('\n Megatonnes of CO2equivalent greenhouse gases did Mauritius emit in 2020')
Mauritius = df[df['geoAreaName'] == 'Mauritius']
print(Mauritius[Mauritius['timePeriodStart'] == 2022]['EN_ATM_GHGT_WLU'])

print('\n country that had the highest percentage of renewable energy in 2023? What was this percentage?')
data_2023 = df[df['timePeriodStart'] == 2023]
print(data_2023[data_2023['EG_FEC_RNEW'] == data_2023['EG_FEC_RNEW'].max()])
```
