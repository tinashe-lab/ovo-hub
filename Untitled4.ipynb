```python
(
ggplot(data=penguins, mapping=aes(x='body_mass_g', y='bill_length_mm', colour="species")) +
geom_histogram(alpha=0.5) +
labs(title='Boxplot of Bill Length by Species')
)

#when you change the aesthetic to a different variable the shape of the histogram changes
#the bin argument removes rows that have no finite values
```

```python
(
ggplot(data=penguins, mapping=aes(x='flipper_length_mm', y='species')) +
geom_density(alpha=0.5) +
labs(title='Graph of flipper Length by Species')
)
```
